```python
from langsmith import Client
```

### Attch feedback to a Specific trace in Langsmith

```python
client=Client()
```

```python
RUN_ID="01a01e5f-7860-7612-b9a4-086dd3720cde"
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="feedback",
    score=0,
    feedback_score_type="api",
)
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="feedback",
    score=0,
    value="This feedback is done by Anurag Yadav",
    feedback_score_type="api",
)
```
