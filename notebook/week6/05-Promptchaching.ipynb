```python
raw_response_2.usage
```

```python
def print_usage_cost(raw_response):
    input_price = 3 / 1_000_000
    output_price = 12 / 1_000_000
    cached_price = 0.75 / 1_000_000

    usage = raw_response.usage

    prompt_tokens = usage.prompt_tokens
    completion_tokens = usage.completion_tokens
    cached_tokens = usage.prompt_tokens_details.cached_tokens or 0
    uncached_prompt_tokens = prompt_tokens - cached_tokens

    prompt_cost = uncached_prompt_tokens * input_price
    cached_cost = cached_tokens * cached_price
    completion_cost = completion_tokens * output_price
    total_cost = prompt_cost + cached_cost + completion_cost

    print(f"Input tokens:  {uncached_prompt_tokens:,} | Cost: ${prompt_cost:.8f}")
    print(f"Cached tokens: {cached_tokens:,} | Cost: ${cached_cost:.8f}")
    print(f"Output tokens: {completion_tokens:,} | Cost: ${completion_cost:.8f}")
    print(f"Total tokens:  {usage.total_tokens:,} | Cost: ${total_cost:.8f}")
```

```python
print_usage_cost(raw_response_2)
```
